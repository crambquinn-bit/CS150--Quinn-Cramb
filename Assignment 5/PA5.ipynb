## Task 1: Weekend Activities as Sets
Ask the user for Saturday and Sunday activities, store them in sets, and display activities completed on both days and on at least one day.

```python

actSat = set()
actSun = set()

# Saturday activities
while True:
    activity = input("Enter an activity from last Saturday (Done to stop): ")

    if activity == "Done":
        break

    actSat.add(activity)

# Sunday activities
while True:
    activity = input("Enter an activity from last Sunday (Done to stop): ")

    if activity == "Done":
        break

    actSun.add(activity)

# Print both sets
print("Saturday activities:", actSat)
print("Sunday activities:", actSun)

# Activities done on both days
print("Activities done on both days:", actSat.intersection(actSun))

# Activities done on at least one day
print("Activities done on at least one day:", actSat.union(actSun))
```

## Task 2: Triangle Calculations
Create a function that calculates the hypotenuse, perimeter, and area of a right triangle using the lengths of its two legs.

```python

def triangle_solve(leg1, leg2):
    hypotenuse = (leg1 ** 2 + leg2 ** 2) ** 0.5
    perimeter = leg1 + leg2 + hypotenuse
    area = (leg1 * leg2) / 2

    return (hypotenuse, perimeter, area)


# Test the function
print(triangle_solve(3, 4))
```

## Task 3: Calculator Class
Create a Calculator class that performs addition, subtraction, multiplication, division, and exponentiation. Track the previous result and total number of calculations.

```python

class Calculator:

    def __init__(self):
        self.prev_result = None
        self.total_calculations = 0

    def add(self, a, b):
        result = a + b
        self.prev_result = result
        self.total_calculations += 1
        return result

    def subtract(self, a, b):
        result = a - b
        self.prev_result = result
        self.total_calculations += 1
        return result

    def multiply(self, a, b):
        result = a * b
        self.prev_result = result
        self.total_calculations += 1
        return result

    def divide(self, a, b):
        if b == 0:
            print("Cannot divide by zero.")
            return None

        result = a / b
        self.prev_result = result
        self.total_calculations += 1
        return result

    def power(self, a, b):
        result = a ** b
        self.prev_result = result
        self.total_calculations += 1
        return result

    def info(self):
        print("Previous result:", self.prev_result)
        print("Total calculations:", self.total_calculations)


# Test the calculator
calc = Calculator()

print(calc.add(5, 3))
print(calc.subtract(10, 4))
print(calc.multiply(3, 4))
print(calc.divide(20, 5))
print(calc.power(2, 3))

calc.info()
```
