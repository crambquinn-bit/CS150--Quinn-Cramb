## Task 1: Input Positive Integer and Factors

### Part A: Find all positive factors

```python
number = int(input("Please input a positive integer: "))
for i in range(1, number + 1):
    if number % i == 0:
        print(i, "is a factor of", number)
```

```python
number = int(input("Please input a positive integer: "))
counter = 0
for i in range(1, number + 1):
    if number % i == 0:
        counter += 1
if counter == 2:
    print("Your number is prime")
else:
    print("Your number is not prime")
```

### Part B: Determine whether a number is prime

```python
number = int(input("Please input a positive integer: "))
counter = 0
for i in range(1, number + 1):
    if number % i == 0:
        counter += 1
if counter == 2:
    print("Your number is prime")
else:
    print("Your number is not prime")
```

## Task 2: Language Translator

```python
translations = {"lacrosse": "lacrosse", "hockey": "hockey", "run": "correr", "university": "universidad", "football": "futbol americano", "hello": "hola", "goodbye": "adios", "dog": "perro", "cat": "gato", "house": "casa"}
while True:
    word = input("Enter an English word to translate to Spanish (press Enter to stop): ").lower()
    if word == "":
        break
    if word in translations:
        print("The Spanish word for", word, "is", translations[word])
    else:
        print("The Spanish word for", word, "is not in this dictionary.")
```

## Task 3: Temperature Summaries

Monthly average high and low temperatures for Louisville, Kentucky, in degrees Fahrenheit.

Source: National Weather Service, 1991–2020 Climate Normals.

```python
temperatures = {"january": [44, 28], "february": [48, 31], "march": [58, 39], "april": [70, 49], "may": [78, 59], "june": [86, 67], "july": [89, 71], "august": [88, 70], "september": [82, 62], "october": [71, 50], "november": [58, 39], "december": [47, 32]}
while True:
    month = input("Enter a month to see Louisville temperatures (type quit to stop): ").lower()
    if month == "quit":
        break
    if month in temperatures:
        print("The average high temperature in Louisville for", month, "is", temperatures[month][0], "and the average low is", temperatures[month][1])
    else:
        print("Not a valid month")
```

## Task 4: Inventory Dictionary and Prices

```python
inventory = {"hammer": [15, 12.49], "wrench": [20, 7.99], "screwdriver": [40, 3.99], "nailgun": [10, 36.59], "padlock": [25, 5.99], "roll_of_duct_tape": [30, 4.95]}

print("1. Check item availability and price")
print("2. Find items within a budget")
print("3. Find total quantity of items")
print("4. Find total value of inventory")

choice = int(input("Enter your choice (1-4): "))
if choice == 1:
    item = input("Enter an item name: ").lower()
    if item in inventory:
        print("Quantity available:", inventory[item][0])
        print("Price: $", inventory[item][1])
    else:
        print("Item is not available.")
elif choice == 2:
    budget = float(input("Enter your maximum budget: "))

    for item in inventory:
        if inventory[item][1] < budget:
            print(item, "$", inventory[item][1])
elif choice == 3:
    total_quantity = 0
    for item in inventory:
        total_quantity += inventory[item][0]

    print("Total quantity of items:", total_quantity)

elif choice == 4:
    total_price = 0
    for item in inventory:
        total_price += inventory[item][0] * inventory[item][1]
    print("Total value of inventory: $", round(total_price, 2))

else:
    print("Invalid choice")
```
