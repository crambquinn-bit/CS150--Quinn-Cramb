```python
##### import random


# Generate a random four-digit secret code
def make_secret():
    secret = ""

    for i in range(4):
        digit = random.randint(0, 9)
        secret += str(digit)

    return secret


# Compare the player's guess with the secret code
def check_guess(secret, guess):
    correct_position = 0
    remaining_secret = []
    remaining_guess = []

    # First, count digits in the correct position
    for i in range(4):
        if secret[i] == guess[i]:
            correct_position += 1
        else:
            remaining_secret.append(secret[i])
            remaining_guess.append(guess[i])

    # Next, count correct digits in the wrong position
    wrong_position = 0

    for digit in remaining_guess:
        if digit in remaining_secret:
            wrong_position += 1
            remaining_secret.remove(digit)

    return correct_position, wrong_position


# Ask the player for a valid guess
def get_guess():
    while True:
        guess = input("Enter your 4-digit guess: ")

        if len(guess) == 4 and guess.isdigit():
            return guess
        else:
            print("Invalid guess. Please enter exactly 4 digits.")


# Play one complete game
def play_game():
    secret = make_secret()
    max_guesses = 10

    print()
    print("I have generated a secret 4-digit code.")
    print("The digits can be between 0 and 9.")
    print("Digits can repeat.")
    print("You have 10 attempts to guess the code.")
    print()

    for attempt in range(1, max_guesses + 1):
        print("Attempt", attempt, "of", max_guesses)

        guess = get_guess()

        correct, wrong = check_guess(secret, guess)

        print("Correct digits in the correct position:", correct)
        print("Correct digits in the wrong position:", wrong)
        print()

        if correct == 4:
            print("Congratulations! You won!")
            print("You guessed the code in", attempt, "attempts.")
            return True

    print("You lost! You used all 10 attempts.")
    print("The secret code was:", secret)
    return False


# Run the game and track wins and losses
def main():
    wins = 0
    losses = 0

    print("Welcome to Mastermind!")
    print("Try to guess the computer's secret number.")

    while True:
        result = play_game()

        if result == True:
            wins += 1
        else:
            losses += 1

        print()
        print("Your record:")
        print("Wins:", wins)
        print("Losses:", losses)
        print()

        while True:
            again = input("Would you like to play again? (yes/no): ").lower()

            if again == "yes" or again == "y":
                break
            elif again == "no" or again == "n":
                break
            else:
                print("Please enter yes or no.")

        if again == "no" or again == "n":
            print("Thanks for playing Mastermind!") 
            break


# Start the game
main()
```
