## Task 1: Types of Triangles

```python
side1= float(input("Enter the first side: "))
side2= float(input("Enter the second side:"))
side3= float(input("Enter the third side:"))

if side1 == side2 and side2 == side3:
    print("The triangle is equilateral.")
elif side1 == side2 or side1 == side3 or side2 == side3:
    print("The triangle is isosceles.")
else:
    print("The triangle is scalene.")
```

## Task 2: Student Grades and At-Risk Students

```python
student_grades= []
for i in range(5):
    grade = float(input("Enter a student grade: "))
    student_grades.append(grade)

average = sum(student_grades) / (len(student_grades))

print("Class average:", average)
print("Minimum grade:", min(student_grades))
print("Maximum grade:", max(student_grades))   

counter = 0

for grade in student_grades:
    if grade < 65:
        counter +=1

print(counter, "students are at risk in the course.")
```

## Task 3: Ranges and Multiples

```python
mult_of_2 = list(range(2,101,2))
mult_of_3 = list(range(3,101,3))

mult_of_both = []

for number in mult_of_2:
    if number in mult_of_3:
        mult_of_both.append(number)

mult_of_either = []

for number in range(1,101):
    if number in mult_of_2 or number in mult_of_3:
        mult_of_either.append(number)

print("Multiples of 2:", mult_of_2)
print("Multiples of 3:", mult_of_3)
print("Multiples of both:", mult_of_both)
print("Multiples of either:", mult_of_either)
```

## Task 4: Shopping List

```python
shopping_list = []

for i in range(7):
    item = input("Enter an item or type done to finish: ")

    if item.lower() == "done":
        break

    shopping_list.append(item)

print("Total number of items:", len(shopping_list))

print("Reverse order:")
print(shopping_list[::-1])

print("Alphabetical order:")
print(sorted(shopping_list))

duplicate = False

for i in range(len(shopping_list)):
    for j in range(i+ 1, len(shopping_list)):
        if shopping_list[i] == shopping_list[j]:
            duplicate = True

if duplicate:
    print("At least one item appears twice in the shopping list.")
else:
    print("No items appear twice in the shopping list.")
```
